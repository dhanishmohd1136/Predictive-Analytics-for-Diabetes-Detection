```python
import numpy as np 
import pandas as pd

df = pd.read_csv('../data/diabetes.csv')
```

```python
df_clean = df.copy()
```

```python
zero_not_valid = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]
print('Colums with medically invalid zero values\n ========================================') 

print(zero_not_valid)
```

```python
print('Count zeros in Inpermissable columns')
print('===============================================')

for col in zero_not_valid:
    zero_count = (df_clean[col]==0).sum()
    zero_percentage = (zero_count / len(df_clean)) * 100

    print(
        f"{col:<15}"
        f" Count: {zero_count:<5}"
        f" Percentage: {zero_percentage:.2f}%"
    )
```

```python
for col in zero_not_valid:
    df_clean[col] = df_clean[col].replace(0,np.nan)
```

```python
missing_values = df_clean.isnull().sum()

missing_report = pd.DataFrame({
    "Missing Count": missing_values,
    "Missing Percentage": (missing_values / len(df_clean)) * 100
})

missing_report = missing_report.sort_values(
    by="Missing Percentage",
    ascending=False
)

print("\nMissing Values After Replacing Zeros")
display(missing_report)
```

```python
print('Median Values Used for Imputation')
print('===================================')

for col in zero_not_valid:
    median_value = df_clean[col].median()

    print(f'{col:<25} Median: {median_value:.1f}')

    
    df_clean[col] = df_clean[col].fillna(median_value)
```

```python
print("Missing Values After Imputation")
print('==================================')
print(df_clean.isnull().sum())

print(
    "\nTotal Missing Values:",
    df_clean.isnull().sum().sum()
)
```

```python
print('Original shape : ',df.shape)
print('Cleaned Shape  : ',df_clean.shape)
```

```python
duplicate_count = df_clean.duplicated().sum()

print("\nDuplicate Rows:", duplicate_count)
```

```python
import os
import shutil


output_path = '../data/processed/cleaned_data.csv'


if os.path.isdir(output_path):
    shutil.rmtree(output_path)


os.makedirs('../data/processed', exist_ok=True)


df_clean.to_csv(output_path, index=False)

```
