```python
import os
import joblib
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
os.makedirs("../data/processed", exist_ok=True)
os.makedirs("../models", exist_ok=True)
```

```python
file_path = "../data/processed/cleaned_data.csv"
df = pd.read_csv(file_path)

print("Dataset Loaded Successfully")
print("Shape:", df.shape)

display(df.head())
```

```python
target_column = "Outcome"

X = df.drop(columns=[target_column])
y = df[target_column]

print("Feature Matrix Shape :", X.shape)
print("Target Vector Shape  :", y.shape)

print("\nFeature Columns:")
print(X.columns.tolist())
```

```python
numerical_columns = X.select_dtypes(
    include=[np.number]
).columns.tolist()

print("\nNumerical Columns:")
print(numerical_columns)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTrain-Test Split Completed")
print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[numerical_columns] = scaler.fit_transform(
    X_train[numerical_columns]
)

X_test_scaled[numerical_columns] = scaler.transform(
    X_test[numerical_columns]
)

print("\nFeature Scaling Completed")
```

```python
joblib.dump(
    scaler,
    "../models/scaler.pkl"
)

print("Scaler saved successfully.")
print("Saved to: ../models/scaler.pkl")
```

```python
joblib.dump(
    scaler,
    "../models/scaler.pkl"
)
```

```python
X_train.to_csv(
    "../data/processed/X_train.csv",
    index=False
)

X_test.to_csv(
    "../data/processed/X_test.csv",
    index=False
)

y_train.to_frame(name=target_column).to_csv(
    "../data/processed/y_train.csv",
    index=False
)

y_test.to_frame(name=target_column).to_csv(
    "../data/processed/y_test.csv",
    index=False
)

print("Unscaled train-test datasets saved.")
```

```python
X_train_scaled.to_csv(
    "../data/processed/X_train_scaled.csv",
    index=False
)

X_test_scaled.to_csv(
    "../data/processed/X_test_scaled.csv",
    index=False
)

print("Scaled feature datasets saved.")
```

```python
train_df = X_train.copy()
train_df[target_column] = y_train.values

test_df = X_test.copy()
test_df[target_column] = y_test.values

train_df.to_csv(
    "../data/processed/train.csv",
    index=False
)

test_df.to_csv(
    "../data/processed/test.csv",
    index=False
)

print("Combined train and test datasets saved.")
```

```python
feature_metadata = pd.DataFrame({
    "Feature": X.columns,
    "DataType": X.dtypes.astype(str).values
})

feature_metadata.to_csv(
    "../data/processed/feature_metadata.csv",
    index=False
)

print("Feature metadata saved.")
```
